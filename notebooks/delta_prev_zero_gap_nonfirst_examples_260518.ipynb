# Non-first `delta_prev_seconds == 0` examples

This notebook shows session examples where `event_idx > 0` and `delta_prev_seconds == 0`, so the event is not the first event but has the same timestamp as the previous event.

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path('data/processed/bpi2012_complete_only/events_encoded_time_features.csv')
df = pd.read_csv(DATA_PATH)
print('rows:', len(df))
print('columns:', df.columns.tolist())
```

```python
zero_gap_nonfirst = df[(df['event_idx'] > 0) & (df['delta_prev_seconds'] == 0)].copy()
print('non-first zero-gap count:', len(zero_gap_nonfirst))
zero_gap_nonfirst[['case_id', 'event_idx', 'activity', 'lifecycle', 'timestamp', 'delta_prev_seconds', 'delta_start_seconds']].head(10)
```

```python
sample_case_ids = zero_gap_nonfirst['case_id'].drop_duplicates().head(5).tolist()
sample_case_ids
```

```python
for case_id in sample_case_ids:
    print('=' * 100)
    print('case_id:', case_id)
    case_df = df[df['case_id'] == case_id].copy()
    case_df = case_df[['case_id', 'event_idx', 'activity', 'lifecycle', 'timestamp', 'delta_prev_seconds', 'delta_start_seconds']]
    display(case_df)
    print('zero-gap non-first rows in this case:')
    display(case_df[(case_df['event_idx'] > 0) & (case_df['delta_prev_seconds'] == 0)])
```

```python
case_id = sample_case_ids[0]
case_df = df[df['case_id'] == case_id].copy()
idxs = case_df[(case_df['event_idx'] > 0) & (case_df['delta_prev_seconds'] == 0)]['event_idx'].tolist()
print('focus case_id:', case_id)
print('zero-gap non-first event_idx:', idxs)
for idx in idxs:
    print('-' * 100)
    print('window around event_idx =', idx)
    window = case_df[(case_df['event_idx'] >= idx - 2) & (case_df['event_idx'] <= idx + 2)].copy()
    display(window[['case_id', 'event_idx', 'activity', 'lifecycle', 'timestamp', 'delta_prev_seconds', 'delta_start_seconds']])
```
